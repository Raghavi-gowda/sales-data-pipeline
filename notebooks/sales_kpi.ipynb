```python
import pandas as pd

df=pd.read_csv('sales.csv')
print(df.shape)
print(df.head())
```

```python
#total_revenue KPI's

total_revenue=df['Sales'].sum().round(2)
total_orders=df['Sales'].nunique()
avg_order_val=(total_revenue/total_orders).round(2)
print("Total revenue ",total_revenue)
print("Total orders ",total_orders)
print("avg_order_val ",avg_order_val)
```

```python
#revenue by year

yearly=df.groupby('Year')['Sales'].sum().round(2)
print("\n Revenue by year\n",yearly)

```

```python
#YoY Growth

yoy_growth=yearly.pct_change()*100
print("\nYear over Year Growth% \n",yoy_growth.round(2))
```

```python
#Revenue by Region & Category

by_region=df.groupby('Region')['Sales'].sum().round(2).sort_values(ascending=False)
print("\nRevenue by Region :\n",by_region)

by_category=df.groupby('Category')['Sales'].sum().round(2).sort_values(ascending=False)
print("\nRevenue by Category :\n",by_category)
```

```python
#top 10 products

top_products=df.groupby('Product Name')['Sales'].sum().round(2).sort_values(ascending=False).head(10)
print("\ntop_products :\n",top_products)
```

```python
monthly=df.groupby(['Year','Month'])['Sales'].sum().round(2)
print("\nMonthly revenue:\n",monthly)
```
